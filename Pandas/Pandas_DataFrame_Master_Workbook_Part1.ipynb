## Q45 add RatingCategory.

```python
# Write your solution here
movies.head(2)
```

```python
def category(rating):
    if rating >= 8:
        return "Hit"
    elif rating >= 6:
        return "Average"
    else:
        return "Flop"

movies["RatingCategory"] = movies["imdb_rating"].apply(category)
movies
```
