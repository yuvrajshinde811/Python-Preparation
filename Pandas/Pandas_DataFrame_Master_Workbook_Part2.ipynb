# Section Q - Apply()

## Q87. Create a custom function to classify movies by rating.

```python
# Write your solution here
def classify_movies(movies):
    if movies['imdb_rating'] > 8:
        return 'Hit'
    elif movies['imdb_rating'] > 6.5:
        return 'Average'
    else:
        return 'Flop'

movies['Category'] = movies.apply(classify_movies, axis=1)
```

```python
movies.head(2)
```

## Q88. Apply the function on imdb_rating.

```python
# Write your solution here
def box_office(movies):
    if movies['imdb_rating'] > 9:
        return "Blockbaster"
    elif movies['imdb_rating'] > 8:
        return "Housefull"
    else:
        return "Thik thak"

movies['Collection'] = movies.apply(box_office, axis=1)
movies
```

## Q90. Categorize batsman_run into Low/Medium/High.

```python
# Write your solution here
runs
def catrun(runs):
    if runs['batsman_run'] > 4000:
        return "High"
    elif runs['batsman_run'] > 2000:
        return "Medium"
    else:
        return "Low"

runs['category'] = runs.apply(catrun, axis=1)
runs
```
