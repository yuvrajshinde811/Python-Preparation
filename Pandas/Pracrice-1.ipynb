```python
# SECTION A — Creating Series (Easy)
```

```python
# Q1
# Create a Series from the list:

# ["Mumbai","Delhi","Pune","Nagpur","Nashik"]

cities = ["Mumbai","Delhi","Pune","Nagpur","Nashik"]
pd.Series(cities)
```

```python
# Q2

# Create a Series from:

# [10,20,30,40,50]

num = [10,20,30,40,50]
pd.Series(num)
```

```python
# Q3

# Assign custom indexes:

# ["A","B","C","D","E"]

a = ["A","B","C","D","E"]
b = [1, 2, 3, 4, 5]
pd.Series(a, index = b)
```

```python
# Q4

# Create a Series and set name as "Cities".

town = ["Manwat", "Parbhani", "Hamdapur"]
pd.Series(town, name="Cities")
```

```python
# Q5

# Create a Series from dictionary:

# {
#     "India":140,
#     "USA":33,
#     "China":142
# }

countries = {
    "India":140,
    "USA":33,
    "China":142
}
pd.Series(countries)
```

```python
# Q6

# Create a Series of floating values.

num1 = [1.2 , 3.5, 4.2, 5.9, 6.7]
pd.Series(num1)
```

```python
# Q7

# Create a Series of booleans.

bool1 = [True, False, True, False]
pd.Series(bool1)
```

```python
# Q8

# Create a Series with duplicate indexes.

a2 = ["Yuvi", "Yash", "Pranav", "Omkar"]
n1 = [1, 2, 3, 2]
pd.Series(a2, index=n1)
```

```python
# Q9

# Create a Series containing mixed datatypes.

mixed = [True, 23, "Yuvi", 5.2]
pd.Series(mixed)
```

```python
# Q10

# Predict the dtype of:

pd.Series([10,20,30.5])
type(pd.Series([10,20,30.5]))
```
