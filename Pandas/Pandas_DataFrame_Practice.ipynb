## 7. Filtering DataFrames

```python
# movies with imdb rating above 8
high_rated = movies[movies['imdb_rating'] > 8]
high_rated[['title_x', 'imdb_rating']].head()
```

```python
# multiple conditions: high rated AND released after 2015
filtered = movies[(movies['imdb_rating'] > 8) & (movies['year_of_release'] > 2015)]
filtered[['title_x', 'year_of_release', 'imdb_rating']]
```

```python
# matches won by a specific team
csk_wins = matches[matches['WinningTeam'] == 'Chennai Super Kings']
csk_wins[['Season', 'Team1', 'Team2', 'WinningTeam']].head()
```

```python
# filtering using isin()
final_teams = matches[matches['Team1'].isin(['Mumbai Indians', 'Chennai Super Kings'])]
final_teams[['Season', 'Team1', 'Team2']].head()
```

### 9.8 `isnull`, `notnull`, `hasnans`

```python
print(movies['imdb_rating'].isnull().sum())
print(movies['imdb_rating'].notnull().sum())
print(movies['imdb_rating'].hasnans)
```

### 9.9 `dropna`

```python
movies_dropped = movies.dropna(subset=['imdb_rating'])
print('Before:', movies.shape)
print('After :', movies_dropped.shape)
```

### 9.10 `fillna`

```python
movies_filled = movies.copy()
movies_filled['imdb_rating'] = movies_filled['imdb_rating'].fillna(movies_filled['imdb_rating'].mean())
movies_filled['imdb_rating'].isnull().sum()
```

### 9.11 `drop_duplicates()` — `keep` parameter

```python
# duplicate rows based on 'Season' + 'Team1' + 'Team2' combo, just to demonstrate 'keep'
subset_cols = ['Season', 'Team1', 'Team2']

print('keep="first" (default) shape:', matches.drop_duplicates(subset=subset_cols, keep='first').shape)
print('keep="last" shape          :', matches.drop_duplicates(subset=subset_cols, keep='last').shape)
print('keep=False shape           :', matches.drop_duplicates(subset=subset_cols, keep=False).shape)
```

### 9.12 `drop`

```python
# drop a column
movies.drop(columns=['poster_path', 'wiki_link']).head(2)
```

```python
# drop rows by index label
matches.drop(index=[0, 1]).head()
```

### 9.13 `apply`

```python
# apply on a single column - custom function
def rating_category(rating):
    if pd.isna(rating):
        return 'Unknown'
    elif rating >= 8:
        return 'Excellent'
    elif rating >= 6:
        return 'Good'
    else:
        return 'Average'

movies_copy['rating_category'] = movies_copy['imdb_rating'].apply(rating_category)
movies_copy[['title_x', 'imdb_rating', 'rating_category']].head()
```

```python
# apply with lambda across a row (axis=1)
matches['match_summary'] = matches.apply(lambda row: f"{row['Team1']} vs {row['Team2']} ({row['Season']})", axis=1)
matches[['match_summary']].head()
```
