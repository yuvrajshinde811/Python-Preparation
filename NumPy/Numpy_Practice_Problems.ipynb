```python
# Create a 1D array from 1 to 20
import numpy as np
a = np.arange(1,21)
print(a)
```

```python
# Create 10 evenly spaced numbers between 5 and 50
a = np.linspace(5,50,10)
print(a)
```

```python
# Create a 4×4 identity matrix
a = np.identity(4)
print(a)
```

```python
# Create a 3×5 array of zeros
a = np.zeros((3,5))
print(a)
```

```python
# Create a 2×3 array of random integers (1–100)
a = np.random.randint(1,100,6).reshape(2,3)
print(a)
```

```python
a = np.arange(24).reshape(4,6)
print(a)
a.shape
a.size
a.ndim
a.dtype
a.itemsize
```

```python
# Convert an integer array to float
a = np.arange(24).reshape(4,6)
a.astype(np.float32)
print(a)
```

```python
# Convert float array to integer
a = np.arange(24, astype=float).reshape(4,6)
a.astype(np.int32)
print(a)
```

```python
# Add 10 to every element
a = np.arange(24).reshape(4,6)
print(a)
print(a + 10)
```

```python
# Multiply array by 5
a = np.arange(24).reshape(4,6)
print(a)
print(a * 5)
```

```python
# Add two arrays element-wise
a = np.arange(12).reshape(3,4)
b = np.arange(4).reshape(4)
print(a)
print(b)
c = a + b
print(c)
```

```python
# Create array 1–10 and return values > 5
a = np.arange(1,11).reshape(2,5)
a[a > 5]
```

```python
# Replace values < 5 with 0
a = np.arange(1,11).reshape(2,5)
print(a)
a[a < 5] = 0
print(a)
```

```python
# Find dot product of two arrays
a = np.arange(12).reshape(3,4)
b = np.arange(12,24).reshape(4,3)
print(a)
print(b)
print(np.dot(a,b))
```

```python
a = np.random.randint(1,50,(4,4))
print(a)
np.max(a)
np.min(a)
np.sum(a)
np.prod(a)
```

```python
a = np.random.randint(1,50,(4,4))
print(a)
np.mean(a)
np.median(a)
np.std(a)
np.var(a)
```

```python
# Find row-wise sum
np.sum(a, axis=1)
```

```python
# Find column-wise mean
np.mean(a, axis=0)
```

```python
# Round values to nearest integer
np.round(np.random.randint(10).reshape(5,2))
```

```python
# Apply log and exponential
a = np.random.randint(1,50,(4,4))
np.exp(a)
np.log(a)
```

```python
a = np.arange(1,13).reshape(3,4)
print(a)
a[1]# Extract 2nd row
a[:,3]# Extract last column
a[1:,1:3]# Extract submatrix:
```

```python
# Print all elements using `np.nditer`
for i in np.nditer(a):
    print(i,end=" ")
```

```python
# Convert 1D array of 12 elements into 3×4
a = np.arange(12)
print(a)
a = np.arange(12).reshape(3,4)
print(a)
```

```python
# Transpose a matrix
np.transpose(a)
```

```python
# Flatten array using `ravel()`
np.ravel(a)
```

```python
# Stack two arrays vertically
a = np.arange(36).reshape(6,6)
b = np.arange(24).reshape(4,6)
np.vstack((a,b))
```

```python
# Stack horizontally
a = np.arange(36).reshape(4,9)
b = np.arange(24).reshape(4,6)
np.hstack((a,b))
```

```python
# Split array into 2 equal parts (row-wise)
a = np.arange(36).reshape(6,6)
print(a)
np.vsplit(a,2)
```

```python
# Split column-wise
a = np.arange(36).reshape(6,6)
print(a)
np.hsplit(a,2)
```

```python
a = np.arange(12).reshape(4,3)
print(a)
# Select rows 1 and 3 using fancy indexing
a[[1,3]]
# Extract elements at positions (0,1), (1,2), (2,0)
a[[0,1,2],[1,2,0]]
# Use boolean indexing to get values > 5
a[a > 5]
```

```python
# Add scalar 5 to array
a + 5
```

```python
# Add row vector `[1,2,3]` to matrix
a + [1,2,3]
```

```python
# Add column vector to matrix
a + np.array([[9],[9],[9],[9]])
a
```
