## Array Functions

```python
a1 = np.random.random((3,3))
a1 = np.round(a1 * 100)
a1
```

```python
# max/min/sum/prod
np.max(a1)
np.min(a1)
np.sum(a1).item()
# np.prod(a1).item()
```

```python
# we can fetch by column and by row also 
# for column -> 0
# for row -> 1
np.min(a1, axis = 0)
np.max(a1, axis = 1)
np.sum(a1, axis = 1)
```

```python
# mean/median/std/var
np.mean(a1)
np.median(a1)
np.std(a1)
np.var(a1, axis= 0)
```

```python
# we can calculate trignometric functions
np.sin(a1)
```

# NumPy Advanced

```python
import numpy as np
import time
```

```python
a = [i for i in range(10000000)]
b = [i for i in range(10000000, 20000000)]
c = []
start = time.time()
for i in range(len(a)):
    c.append(a[i] + b[i])

print(time.time()-start)
```

```python
a = np.arange(10000000)
b = np.arange(10000000, 20000000)
start = time.time()
c = a + b

print(time.time()-start)
```

## Advance Indexing

```python
# Fancy Indexing
a = np.arange(20).reshape(5,4)
a
```

```python
a[0:,[1,2]]
```

```python
# Boolean indexing
a = np.random.randint(1,100,24).reshape(6,4)
a
```

```python
a > 50 # comapred evry element with 50
a[a>50] # returns all elements which are greater than 50
```

```python
a[(a % 2 == 0) & (a > 50)]
```

```python
a[~(a % 7 == 0)]
```
