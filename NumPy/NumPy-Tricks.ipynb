## np.concatenate()

```python
# concatenate means connecting the two arrays
c = np.arange(6).reshape(2,3)
d = np.arange(6,12).reshape(2,3)

print(c)
print(d)
```

```python
np.concatenate((c,d), axis=1) # rowwise concatenation
```

```python
np.concatenate((c,d), axis=0) # rowwise concatenation
```
